```python
import socket
import time
import pandas as pd

def traceroute(destination, max_hops=15, timeout=2):

    destination_ip = socket.gethostbyname(destination)

    print("Tracing route to:", destination)
    print("Destination IP:", destination_ip)
    print("-" * 55)

    results = []

    for ttl in range(1, max_hops + 1):

        receiver = socket.socket(
            socket.AF_INET,
            socket.SOCK_RAW,
            socket.IPPROTO_ICMP
        )

        sender = socket.socket(
            socket.AF_INET,
            socket.SOCK_DGRAM,
            socket.IPPROTO_UDP
        )

        receiver.settimeout(timeout)
        receiver.bind(("", 33434))

        # Windows-compatible TTL configuration
        sender.setsockopt(
            socket.IPPROTO_IP,
            socket.IP_TTL,
            ttl
        )

        start_time = time.perf_counter()

        try:
            sender.sendto(
                b"Traceroute test",
                (destination_ip, 33434 + ttl)
            )

            _, address = receiver.recvfrom(512)

            elapsed = (time.perf_counter() - start_time) * 1000

            hop_ip = address[0]

            results.append({
                "Hop": ttl,
                "IP Address": hop_ip,
                "Latency (ms)": round(elapsed, 2)
            })

            print(f"{ttl:2}  {hop_ip:18}  {elapsed:.2f} ms")

            if hop_ip == destination_ip:
                print("\nDestination reached!")
                break

        except socket.timeout:
            print(f"{ttl:2}  * Request timed out")

        finally:
            sender.close()
            receiver.close()

    return pd.DataFrame(results)
```

```python
import subprocess
import re
import pandas as pd

def windows_traceroute(destination, max_hops=15):

    command = [
        "tracert",
        "-4",
        "-d",
        "-h", str(max_hops),
        "-w", "1500",
        destination
    ]

    result = subprocess.run(
        command,
        capture_output=True,
        text=True
    )

    print(result.stdout)

    return result.stdout


output = windows_traceroute("google.com", 10)
```

```python
import re
import pandas as pd

def parse_traceroute(output):

    records = []

    for line in output.splitlines():

        match = re.match(
            r"\s*(\d+)\s+(.*)",
            line
        )

        if not match:
            continue

        hop = int(match.group(1))
        remaining = match.group(2)

        ip_match = re.search(
            r"(\d{1,3}(?:\.\d{1,3}){3})",
            remaining
        )

        if ip_match:
            ip = ip_match.group(1)
        else:
            ip = None

        latencies = re.findall(
            r"(\d+)\s*ms",
            remaining
        )

        records.append({
            "Hop": hop,
            "IP Address": ip,
            "Latency 1 (ms)": int(latencies[0]) if len(latencies) > 0 else None,
            "Latency 2 (ms)": int(latencies[1]) if len(latencies) > 1 else None,
            "Latency 3 (ms)": int(latencies[2]) if len(latencies) > 2 else None
        })

    return pd.DataFrame(records)
```

```python
df = parse_traceroute(output)

df
```

```python
latency_columns = [
    "Latency 1 (ms)",
    "Latency 2 (ms)",
    "Latency 3 (ms)"
]

df["Average Latency (ms)"] = df[latency_columns].mean(axis=1)

df.head()
```

```python
import plotly.express as px

valid_data = df.dropna(
    subset=["Average Latency (ms)"]
)

fig = px.line(
    valid_data,
    x="Hop",
    y="Average Latency (ms)",
    markers=True,
    title="Network Latency Across Traceroute Hops",
    hover_data=["IP Address"]
)

fig.show()
```

```python
import requests

def get_ip_location(ip):

    if not ip:
        return None

    try:
        response = requests.get(
            f"https://ipwho.is/{ip}",
            timeout=10
        )

        data = response.json()

        if not data.get("success"):
            return None

        return {
            "IP Address": ip,
            "Country": data.get("country"),
            "City": data.get("city"),
            "Region": data.get("region"),
            "ISP": data.get("connection", {}).get("isp"),
            "Latitude": data.get("latitude"),
            "Longitude": data.get("longitude")
        }

    except requests.RequestException as e:
        print(f"Lookup failed for {ip}: {e}")
        return None
```

```python
locations = []

for ip in df["IP Address"].dropna().unique():

    location = get_ip_location(ip)

    if location:
        locations.append(location)

geo_df = pd.DataFrame(locations)

geo_df
```

```python
final_df = df.merge(
    geo_df,
    on="IP Address",
    how="left"
)

final_df
```

```python
import plotly.express as px

fig = px.scatter_geo(
    geo_df,
    lat="Latitude",
    lon="Longitude",
    hover_name="IP Address",
    hover_data=[
        "Country",
        "City",
        "ISP"
    ],
    color="Country",
    title="Geographical Distribution of Network Hops",
    size_max=15
)

fig.update_geos(
    showcountries=True,
    showland=True,
    fitbounds="locations"
)

fig.show()
```
